## 2. Fields: feature / label / context / excluded

*Sort every field you plan to touch into these four buckets. Excluded needs a why.*

## Fields

### Features
- gsc_impressions
- gsc_clicks
- gsc_avg_position
- ga4_sessions
- scroll_events

These are historical performance metrics that are available before making a content refresh decision.

### Label / Proxy
- Future decline in search performance (to be derived later for modeling).

### Context
- report_date
- client_hash_id
- content_hash_id

These identify when and where the observations come from but are not direct prediction features.

### Excluded
- Any future or label-derived information.

Reason: Using future information would cause data leakage and lead to unrealistically high model performance.

```python
from datasets import load_dataset
from google.colab import userdata

HF_TOKEN = userdata.get("HF_tokke")

dataset = load_dataset(
    "FlyRank/internship-warehouse",
    "fact_content_daily_performance",
    token=HF_TOKEN,
    split="train[:100000]"   # only first 100k rows
)

df = dataset.to_pandas()

print(df.shape)
print(df.head())
```

```python
df.head()
```

```python
import pandas as pd

df["report_date"] = pd.to_datetime(df["report_date"])

month = "2025-03"

march = df[df["report_date"].dt.strftime("%Y-%m") == month]

print("Rows:", len(march))
print("Start:", march["report_date"].min())
print("End:", march["report_date"].max())
```

```python
available = march[
    (march["gsc_data_available"] == True) &
    (march["ga4_data_available"] == True)
]

print("Rows with both GSC and GA4 available:", len(available))
```

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

## Data Limits

- The dataset measures website performance metrics but does not explain the exact reasons behind performance changes.
- The current sample contains Google Search Console data, but Google Analytics 4 data is not available, so user engagement analysis is limited.
- External factors such as Google algorithm updates, competitor actions, or seasonal trends are not captured.
- The loaded sample only covers part of the timeline (up to March 2025), so it does not represent the complete warehouse history.
- This dataset is useful for decision support but should not be treated as proof of cause and effect.
